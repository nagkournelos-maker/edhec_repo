```python
# Construct the file path using os.path.join
desktop_path = os.path.join(os.path.expanduser("~"), "Desktop")
file_path = os.path.join(desktop_path, "data", "Portfolios_Formed_on_ME_monthly_EW.csv")

# Read the CSV file into a DataFrame
prices = pd.read_csv(file_path , header=0 , index_col=0 , parse_dates =True , na_values=-99.99)
```

```python
prices.head()
```

```python
returns = prices
columns = ['Lo 10', 'Hi 10']
returns = returns[columns]
returns = returns/100
returns.head()
```

```python
volatility= returns.std()
volatility
```

```python
annualized_vol = volatility*np.sqrt(12)
annualized_vol
```

```python
n_months = returns.shape[0]
return_per_month = ((returns+1).prod()**(1/n_months)) - 1
return_per_month
```

```python
annualized_ret = ((return_per_month+1)**12) -1
```

```python
annualized_ret
```

```python
ret_risk_ratio = annualized_ret/annualized_vol
riskfree = 0.03
an_excess_ret= annualized_ret - riskfree
print(ret_risk_ratio)
sharpe_ratio = an_excess_ret/annualized_vol
sharpe_ratio
```
