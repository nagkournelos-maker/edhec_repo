```python
%load_ext autoreload
%autoreload 2
%matplotlib inline
import os
import pandas as pd
import numpy as np
import edhec_risk_kit as erk
```

```python
ind_return= erk.get_ind30_returns()
ind_size = erk.get_ind30_size()
ind_nfirms= erk.get_ind30_nfirms()
```

```python
ind_size.sum(axis='columns')
```

```python
ind_size.shape
```

```python
ind_nfirms.shape
```

```python
ind_mktcap = ind_nfirms * ind_size
```

```python
total_mktcap = ind_mktcap.sum(axis='columns')
```

```python
total_mktcap.plot()
```

```python
ind_capweight = ind_mktcap.divide(total_mktcap , axis='rows')
```

```python
ind_capweight.head()
```

```python
ind_capweight['1926'].sum(axis='columns')
```

```python
ind_capweight[['Fin','Steel']].plot()
```

```python
total_market_return = (ind_capweight*ind_return).sum(axis='columns')
```

```python
total_market_index = erk.drawdown_compute(total_market_return)['wealth index']
```

```python
total_market_index.plot()
```

```python
total_market_index['1980':].plot(figsize=(12,6))
total_market_index['1980':].rolling(window=36).mean().plot()
```

```python
#trailing 36 month returns
tmi_tr36rets = total_market_return.rolling(window=36).aggregate(erk.annualized_return,periods_per_year=12)
```

```python
tmi_tr36rets.plot(figsize=(12,6),label='trailing 36 month')
total_market_return.plot(label='returns')
```

### rolling correlation , multi-indexes .groupby

```python
#time series corellation
ts_corr = ind_return.rolling(window=36).corr()
```

```python
ts_corr.tail()
```

```python
ts_corr.index.names = ['date','industry']
```

```python
ts_corr.tail()
```

```python
ind_tr36corr = ts_corr.groupby(level='date').apply(lambda cormat: cormat.values.mean())
```

```python
ind_tr36corr.plot()
```

```python
tmi_tr36rets['2007':].plot(label='tr36 mo rets' , figsize=(12,6),legend=True)
ind_tr36corr['2007':].plot(label='tr36 mo corr', secondary_y=True,legend=True)
```

```python
tmi_tr36rets.corr(ind_tr36corr)
```
